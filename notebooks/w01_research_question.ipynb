1) My lane & why:
I opt for "lane 2: Refresh / Content Opportunity Scoring".
A content inventory decays very rapidly as search trend data changes, search engine result page (SERPs) display different types of results, and competitor’s content evolve. Manually going through many thousands of pages to determine which piece to grow, prune, or update is an unsustainable effort. Leverage machine learning to order the list of pages means that rather than a shot in the dark human process of reviewing pages, we have a decision support, evidence-based process so editors use their time to edit pages in highest demand with decreasing search visibility.

2) The Question: Decision, Action, and the Cost of a Wrong Call
The Framing
Search Question: Which high-visibility content items are displaying observable signals of performance decay and should be prioritized for review?
Unit of Analysis: A single pseudonymized content item (page) over a trailing time window.
The Decision: Determining the exact sequence in which content editors should evaluate the production queue.
The Action: A human reviewer opens the top-ranked page, inspects the associated reason codes, and executes a targeted content update (e.g., expansion, metadata revision, or structural overhaul).

The Cost Matrix
False Positive (Type I Error): The model flags a healthy page as "declining".
 Operational Cost: Wasted editorial hours spent analyzing and modifying a page that did not require intervention.
False Negative (Type II Error): The model misses a high-demand page that is actively decaying.
 Business Cost: Lasting loss of organic search traffic, impressions, and client user engagement to competing domains.

3) Quick Look at the Data (2-3 Real Numbers)

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

total_rows = len(df)
unique_content = df['content_id'].nunique()

declining_count = (df['trend_direction'] == 'down').sum()
pct_declining = (declining_count / total_rows) * 100

high_demand_decline = df[(df['impressions_90d'] >= 500) & (df['trend_direction'] == 'down')].shape[0]

print(f"Number 1: Dataset contains {total_rows} rows at a clean 1:1 grain of {unique_content} unique content items.")
print(f"Number 2: {pct_declining:.2f}% of the content library ({declining_count} pages) is classified as declining.")
print(f"Number 3: There are {high_demand_decline} high-demand pages (>=500 impressions) actively losing momentum.")
```

 The above data pull shows that over a quarter of the dataset is in active decline, but what's more interesting is that there are hundreds of high-demand pages in active decline– these represent thousands of organic visits per week which is an immediate problem that we will be failing to manage effectively without it. A prioritized learn list would ensure we don’t fail at the manual triage step on these high priority pages.

4) Careful Words: What I Can and Can't Claim
To maintain scientific integrity throughout this capstone, we establish clear guardrails regarding what this project can realistically prove.

What we CAN claim: We can claim that our model surfaces historical, observable search signals (such as impression drops, position decay, and freshness changes) to rank pages by their statistical risk of decline, beating a simple rule-based baseline.

What we CANNOT claim: We cannot claim that an item's decline is caused by a specific Google algorithm factor. Furthermore, because this is an observational dataset, we cannot guarantee that a manual content refresh will definitively cause a recovery in search traffic. This remains a decision-support ranking tool, not a causal guarantee.
